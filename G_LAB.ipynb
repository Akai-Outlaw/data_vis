```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
```

```python
x = np.arange(1, 12 , 2)
y = np.arange(2, 13, 2)

fig, ax = plt.subplots(figsize=(6,4))

ax.plot(x, y, linestyle="-.", color="#003049", marker="o", markersize="10")
plt.show()
```

```python

plt.plot(x, y, linestyle="-.", marker="*")
plt.show()
```

```python
plt.plot(x, y, linestyle=":", marker="*", markersize="15")
plt.show()
```

```python
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 170, 190, 160, 300, 310, 320, 330])
```

```python
# create a lineplot, custom color, dotted linestyle, diamond markers
fig, ax = plt.subplots()

ax.plot(x, y,
linestyle=":", 
color="#0f4c5c",
markerfacecolor="#e36414", 
marker="d", 
markersize=10, 
markeredgewidth=1
)
plt.title("Sports Watch Data")
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")
plt.show()
```

```python
# create a lineplot, custom color, dotted linestyle, diamond markers
fig, ax = plt.subplots()

ax.plot(x, y,
linestyle=":", 
color=(0.2, 0.4, 0.8),
markerfacecolor=(0.5,0.8, 0.4), 
marker="d", 
markersize=10, 
markeredgewidth=1
)
plt.title("Sports Watch Data", color=(0.8, 0.6, 0.5))
plt.xlabel("Average Pulse", color = (0.3, 0.2, 0.7))
plt.ylabel("Calorie Burnage", color= (0.5,0.5, 0.9))
plt.show()
```

```python
years=[1970,1980,1985,1990,1995,2000,2005,2010,2015,2020,2025,2030,2035]
pops=[3.6,4.0,4.4,4.8,5.3,5.7,6.1,6.5,7.3,7.3,7.3,7.4,7.5]
```

```python
# title of world pop
# labels growth by year, pop in billions green red
plt.plot(years, pops, 
linestyle="--", 
marker="v",
color="#005f73",
markerfacecolor="#e63946",
markersize=8
)
plt.title("World Population", color="Blue")
plt.ylabel("Populations in Billions", color="green")
plt.xlabel("Growth By Years", color="red")
plt.show()
```
